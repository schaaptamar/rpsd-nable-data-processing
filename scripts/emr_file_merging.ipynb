```python
# merging by "cycleid" in cycle_event_df and "id" in cycle_df
merged_df = pd.merge(cycle_event_df, cycle_df, left_on="cycleid", right_on="id", how="left")
```

## Filtering dataset

1) cases that do not have height/weight included
2) cases in which cancellation was immediately after stim start (not enough time to determine poor response to stim)

```python
# There are plenty of test records in nAble - maybe ask Grant to filter those out. For now, filtering to get rid of NA in weight and height will get rid of those cases
merged_df = merged_df[merged_df["weight"].notna() & merged_df["height"].notna()]

print(merged_df.shape)
```

```python
# Deleting rows where the cyclestart and the cycleend are less than or equal to 2 days apart
merged_df["cyclestart"] = pd.to_datetime(merged_df["cyclestart"])
merged_df["cycleend"]   = pd.to_datetime(merged_df["cycleend"])

merged_df["duration"] = (merged_df["cycleend"] - merged_df["cyclestart"]).dt.days

# cycles are only cancelled for poor response at at least 5-6 days in and completed cycles don't trigger before day 8, so getting rid of cycles less than 5 days
merged_df = merged_df[merged_df["duration"] >= 5]
```

```python
print(merged_df.shape)
```

```python
# Getting rid of rows with NA in "sart_id" since those are not IVF cycles (or they are IVF cycles that are unfinished)
merged_df = merged_df.dropna(subset=["sart_id"])
```

```python
print(merged_df.shape)
```

```python
# sort all values of "duration" from smallest to largest
merged_df = merged_df.sort_values(by="duration")
```

```python
#manual entering of missing dates:
# if cycleid = 63799a3d-c145-40b5-9234-b2369dacb25f, then retrieval_date = 2021-11-09

# if cycleid = "4206ba52-6b2a-4037-acb4-4c9cff348d31", then retrieval_date = 2022-04-26

# if sart_id = "n_2478", then response_type = "patient_dropped"

# if sart_id = "n_266", then retrieval_date = 2021-12-16

# if sart_id = "n_99", then retrieval_date = 2021-11-16

# if sart_id = "271", then retrieval_date = 2021-12-17 (patid 23482)
```

```python
# converting trigger_date and retrieval_date to datetime
merged_df["trigger_date"] = pd.to_datetime(merged_df["trigger_date"], errors="coerce")
merged_df["retrieval_date"] = pd.to_datetime(merged_df["retrieval_date"], errors="coerce")

# if status == "Completed" and retrieval_date is NA, then retrieval_date = trigger_date + 2 days
merged_df.loc[(merged_df["status"] == "Completed") & (merged_df["retrieval_date"].isna()), "retrieval_date"] = merged_df["trigger_date"] + pd.Timedelta(days=2)

# if status == "Completed" and trigger_date is NA, then trigger_date = retrieval_date - 2 days
merged_df.loc[(merged_df["status"] == "Completed") & (merged_df["trigger_date"].isna()), "trigger_date"] = merged_df["retrieval_date"] - pd.Timedelta(days=2)
```
