## Merging dfs

Calling in data from: 1)emr cycle merged data, 2) emr aspirations data, and 3) lab ivf data


```python
# --- emr_cycle_merged import ---
cycle_merged_PATH = Path("../data/emr_file_merging_output.csv")

# loading data
cycle_merged_df = pd.read_csv(cycle_merged_PATH)

# in cycle_merged_df, get rid of duplications of full rows, keeping only the first occurrence
cycle_merged_df = cycle_merged_df.drop_duplicates(keep='first')

# --- emr_aspirationst ---
aspirations_PATH = Path("../data/emr_cycle_aspirations_filtered.csv")

# loading data
aspirations_df = pd.read_csv(aspirations_PATH)

# in aspirations_df, get rid of duplications of full rows, keeping only the first occurrence
aspirations_df = aspirations_df.drop_duplicates(keep='first')

# --- lab_ivf import ---
lab_ivf_PATH = Path("../data/ivf_lab_data_cleaned.csv")

# loading data
lab_ivf_df = pd.read_csv(lab_ivf_PATH)

# in lab_ivf_df, get rid of duplications of full rows, keeping only the first occurrence
lab_ivf_df = lab_ivf_df.drop_duplicates(keep='first')
```

```python
display(lab_ivf_df.head())
```

```python
# in aspirations_df, find and display non-NA duplicates of accession_number
duplicates = aspirations_df[aspirations_df['accession_number'].notnull() & aspirations_df.duplicated(subset='accession_number', keep=False)]
display(duplicates)
duplicates.shape
```

```python
# take the leading "#" out of lab_ivf_df's "Cycle #" column
lab_ivf_df['Cycle #'] = lab_ivf_df['Cycle #'].str.lstrip('#')

# now merging lab_ivf_df by "Cycle #" with aspirations_df by "accession_number", only keeping rows that are present in both dataframes
merged_df = pd.merge(lab_ivf_df, aspirations_df, left_on='Cycle #', right_on='accession_number', how='inner')

## drop accession_number column from merged_df
#merged_df.drop(columns=['accession_number'], inplace=True)
```

```python
# now merging merged_df by "cycleid" with cycle_merged_df by "cycleid", only keeping rows that are present in both dataframes
merged_df = pd.merge(merged_df, cycle_merged_df, left_on='cycleid', right_on='cycleid', how='inner')
```

## Adjusting data to look at proportions of usable embryos

```python
df = merged_df.copy()
```

```python
# strip all column names of leading and trailing whitespace
df.columns = df.columns.str.strip()
```

```python
# fill in missing values for "# of D5 usable blast", "# of D6 usable blast", and "# of D7 usable blast" with 0, then create a new column "usable_embryos" that sums these three columns
df["usable_embryos"] = (
    df["# of D5 usable blast"].fillna(0)
    + df["# of D6 usable blast"].fillna(0)
    + df["# of D7 usable blast"].fillna(0)
    + df["D5 Normal G"].fillna(0)
    + df["D5 Normal F"].fillna(0)
    + df["D5 Normal P"].fillna(0)
    + df["D5 Abnormal G"].fillna(0)
    + df["D5 Abnormal F"].fillna(0)
    + df["D5 Abnormal P"].fillna(0)
    + df["D6 Normal G"].fillna(0)
    + df["D6 Normal F"].fillna(0)
    + df["D6 Normal P"].fillna(0)
    + df["D6 Abnormal G"].fillna(0)
    + df["D6 Abnormal F"].fillna(0)
    + df["D6 Abnormal P"].fillna(0)
    + df["D7 Normal G"].fillna(0)
    + df["D7 Normal F"].fillna(0)
    + df["D7 Normal P"].fillna(0)
    + df["D7 Abnormal G"].fillna(0)
    + df["D7 Abnormal F"].fillna(0)
    + df["D7 Abnormal P"].fillna(0)
    + df["# 2PN"].fillna(0)
)

# calculate usable embryos per retrieved oocyte and usable embryos per mature retrieved oocyte, filling in missing values with 0
df["usable_euploid_per_retrieved"] = (
    (df["D5 Normal G"].fillna(0) 
     + df["D6 Normal G"].fillna(0) 
     + df["D7 Normal G"].fillna(0) 
     + df["D5 Normal F"].fillna(0) 
     + df["D6 Normal F"].fillna(0) 
     + df["D7 Normal F"].fillna(0))/
    df["# Egg retrieved/Thawed"].fillna(0)
)

df["usable_euploid_per_mature_retrieved"] = (
    (df["D5 Normal G"].fillna(0) 
     + df["D6 Normal G"].fillna(0) 
     + df["D7 Normal G"].fillna(0) 
     + df["D5 Normal F"].fillna(0) 
     + df["D6 Normal F"].fillna(0) 
     + df["D7 Normal F"].fillna(0))/
    df["# Mature eggs"].fillna(0)
)

# Getting totals of D5, D6, and D7 normal embryos for each cycle
df["euploid_d5"] = (
    df["D5 Normal G"].fillna(0)
    + df["D5 Normal F"].fillna(0)
    + df["D5 Normal P"].fillna(0)
)

df["euploid_d6"] = (
    df["D6 Normal G"].fillna(0)
    + df["D6 Normal F"].fillna(0)
    + df["D6 Normal P"].fillna(0)
)

df["euploid_d7"] = (
    df["D7 Normal G"].fillna(0)
    + df["D7 Normal F"].fillna(0)
    + df["D7 Normal P"].fillna(0)
)

# Filling in trigger date and retrieval date with the values from the other dataframe if they are missing in one of them

# Convert dates to datetime
df["ER Date"] = pd.to_datetime(df["ER Date"], format="%m/%d/%Y", errors="coerce")
df["retrieval_date"] = pd.to_datetime(df["retrieval_date"], errors="coerce")
df["trigger_date"] = pd.to_datetime(df["trigger_date"], errors="coerce")

# Fill missing retrieval dates from ER Date
df["retrieval_date"] = df["retrieval_date"].fillna(df["ER Date"])

# If trigger date is missing, assume 2 days before retrieval
df["trigger_date"] = df["retrieval_date"] - pd.Timedelta(days=2)

# If retrieval date is missing, assume 2 days after trigger
df["retrieval_date"] = df["trigger_date"] + pd.Timedelta(days=2)


"""
# Convert dates to datetime
df["ER Date"] = pd.to_datetime(df["ER Date"], format="%m/%d/%Y", errors="coerce")
df["retrieval_date"] = pd.to_datetime(df["retrieval_date"], errors="coerce")
df["trigger_date"] = pd.to_datetime(df["trigger_date"], errors="coerce")

# Fill missing retrieval dates from ER Date
df["retrieval_date"] = df["retrieval_date"].fillna(df["ER Date"])

# If trigger date is missing, assume 2 days before retrieval
mask = (df["status"] == "Completed") & df["trigger_date"].isna()
df.loc[mask, "trigger_date"] = df.loc[mask, "retrieval_date"] - pd.Timedelta(days=2)

# If retrieval date is missing, assume 2 days after trigger
mask = (df["status"] == "Completed") & df["retrieval_date"].isna()
df.loc[mask, "retrieval_date"] = df.loc[mask, "trigger_date"] + pd.Timedelta(days=2)
"""
```
